```python
import os
from sys import prefix

from PyQt6.QtCore.QFileInfo import suffix
from openai import OpenAI
from dotenv import load_dotenv, find_dotenv
load_dotenv()
_ = os.environ.get('GEMINI_API_KEY')
```

```python
client = OpenAI(
    api_key=_,
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/"
)
```

```python
completion = client.chat.completions.create(
    model='gemini-3.5-flash',
    messages=[
        {'role': 'system', 'content':'you are a helpful assistant'},
        {'role':'user', 'content':'write a haiku about recursion in programming'},
    ]
)
```

```python
print(completion.choices[0].message)
```

```python
examples = [
    {"input": "Hi you are a experienced cricketer", 'output':'cricketer'},
    {"input": "Tell me about off cutter", 'output':'baller'},
]
```

```python
from langchain_core.prompts import FewShotPromptTemplate
```

```python
example_template = '''
Ticket : {input}
category: {output}
'''
```

```python
few_shot_prompt= FewShotPromptTemplate(
    examples=examples,
    example_prompt=PromptTemplate(input_variables=['input', 'output'], template=example_template),
    prefix="classify",
    suffix="\nTicket : {user_input}\n Category:",
    input_variables=['user_input'],
)
```

```python
print(few_shot_prompt)
```
